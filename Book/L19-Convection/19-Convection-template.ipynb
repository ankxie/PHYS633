## 2. To execute: I will show you how to use a quite nifty function that creates colored lines

```python
def color_lines(ax, x, y, z, min_col, max_col, cmap):
    points = np.array([x, y]).T.reshape(-1, 1, 2)

    segments = np.concatenate([points[:-1], points[1:]], axis=1)

    lc = LineCollection(segments, cmap=plt.get_cmap(cmap),
                    norm=plt.Normalize(min_col, max_col))
    lc.set_array(z)
    lc.set_linewidth(2)

    ax.add_collection(lc)

    return lc
```

Here I create a color map that is called cm_plusmin, by defining 256 RGB values.

This is taken from an excellent compilation of B&W and color-blind friendly color maps: https://sronpersonalpages.nl/~pault/

```python
cols = []
for x in np.linspace(0,1, 256):
    rcol = 0.237 - 2.13*x + 26.92*x**2 - 65.5*x**3 + 63.5*x**4 - 22.36*x**5
    gcol = ((0.572 + 1.524*x - 1.811*x**2)/(1 - 0.291*x + 0.1574*x**2))**2
    bcol = 1/(1.579 - 4.03*x + 12.92*x**2 - 31.4*x**3 + 48.6*x**4 - 23.36*x**5)
    cols.append((rcol, gcol, bcol))

cm_plusmin = cl.LinearSegmentedColormap.from_list("PaulT_plusmin", cols)
```

>1. For each star, where is the condition for instability fulfilled?
>
> The condition for instability is $\nabla_\mathrm{rad} > \nabla_\mathrm{ad}$. For a sun-like star, this condition is fulfilled between ~0.7 $R_\odot$ and the surface of the star. For a $10 M_\odot$ star, this condition is fufilled at the core, between the center of the star and ~1 $R_\odot$.
>
>2. For each star, explain the behavior of the real temperature gradient outside of the core (in other words, which term in the equation for $\nabla_\mathrm{rad}$ makes the most difference between a sun-like and high mass star?).
>
> $\nabla_\mathrm{rad}$ is proportional to opacity. Outside of the core, the $10 M_\odot$ star has low opacity. In contrast, there is a region outside of the core of the sun-like star that has high opacity. This causes there to be a region outside of the core of the sun-like star to be convective. Where the star is convective, the real temperature gradient is close to the adiabatic temperature gradient. Where the star is not convective, the real temperature gradient equals the radiative temperature gradient.
>
>3. There is a very famous graph in the Kippenhahn (22.7 in first edition, reproduced in the lecture slides). Based on this graph and the ones you just made, comment on the change in energy transport as a function of stellar mass.  
>
> The structure of convective and radiative zones within a star changes as a function of stellar mass. A $1 M_\odot$ star is mostly radiative, with an outer convective layer. However, once a main sequence star is massive enough, its core will become a convective zone. The size of this central convective zone is proportional to the mass of the star. These massive stars with convective cores don't have a convective outer layer. Meanwhile, stars with masses less than $1 M_\odot$ will have a convective outer layer. The depth this layer reaches increases as mass decreases. The least massive main sequence stars are entirely convective.
