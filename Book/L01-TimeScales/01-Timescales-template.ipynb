## 2. Kelvin-Helmholtz timescale

> **TODO (In class)**: <font color='red'>Write a brief paragraph that describe the calculation we make below. Use Latex formatting to render the math</font>

We have calculated the gravitational potential energy lost during the Sun's lifetime from a cloud to today. The equation is:

$\Delta U_g = \frac{3GM_\odot^2}{5R_\odot}$

Half of the loss gravitational potential energy is radiated away.

```python
##################
## In class

U_g = (3 * G * Msun**2) / 5 / Rsun

print('The potential energy lost by Sun:', U_g.to(u.joule))

t_KH = U_g/2/Lsun

print('The KH timescale:', t_KH.to(u.Myr))
```

> **TODO**: <font color='red'>Please write a small paragraph with an interpretation of the result obtained:</font>

Through carbon dating (and other techniques), we know that the Earth is much older than the KH timescale. Therefore, gravitational contraction cannot sufficiantly explain how the Sun generates energy.

## 4. Hydrodynamic timescale
> **TODO** <font color='red'>: Write a brief paragraph that describe the calculation we make below. Use Latex formatting to render the math</font>

How long would it take for the Sun to collapse if its pressure disappeared? We can modal this scenerio by considering how long it would take a bit of mass at the edge of the Sun to fall to the center of the Sun due to gravity. The gravitational acceleration of this bit of mass due to the enclosed mass is:

$g = \frac{GM_\odot}{R_⊙^2}$

We make an appromation by treating g as constant. By doing this, we are overestimating the amount of time it will take for the Sun to collapse because in reality, g increases as the bit of mass comes closer to the center.

$t_{max} = (\frac{R_\odot}{g})^{0.5} = (\frac{R_\odot^3}{GM_⊙})^{0.5}$

```python
##################
## In class

t_max = ( Rsun ** 3/ G / Msun)**0.5
print("The maximum amount of time it would take for the Sun to collapse is", t_max.to(u.minute))
```

> **TODO**: <font color='red'>Please write a small paragraph with an interpretation of the result obtained:</font>

If pressure disappeared, it would take the Sun 26.5 minutes to collapse. This is a much smaller timescale than the nuclear timescale. This tells us that the timescale at which the Sun readjusts to perturbations is much smaller than its evolutionary timescale. Therefore, we can treat the Sun as a static system.

Stars can be approximated as blackbodies. Blackbody radiation follows Wien's law and the Stefan-Boltzmann law.

Wien's law: $\lambda_{max} = 3 × 10^6 × T^{-1}$

Wien's law relates the wavelength which the star is most luminous with temperature. Therefore, we are able to use the spectrum of a star to estimate its temperature.

Stefan-Boltzmann law: $F = \sigma T^4$

The Stefan-Boltzmann law relates the flux of a star's total radiation with temperature. If we measure the power output of a star (such as by counting the number of photons we collect when we point a telescope at the star), we can estimate its temperature.
